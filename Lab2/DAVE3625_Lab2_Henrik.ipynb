```python
# Import dependencies
import pandas as pd
import numpy as np
import seaborn as sns
from scipy import stats
import matplotlib as plt
```

## Task 1

```python
url = 'data/stud.csv'
```

```python
df_stud = pd.read_csv(url, sep=',')
df_stud.head()
```

```python
df_stud.info()
```

```python
df_stud.isna().sum()
```

```python
df_stud = df_stud.replace(r'^\s*$', np.nan, regex=True)
df_stud.isna().sum()
```

```python
df_stud
```

```python
df_stud['Age'].fillna(0)
df_stud.dropna(inplace=True)
df_stud.isna().sum()
```

```python
df_stud['Age'] = df_stud['Age'].astype(str).astype(int)
df_stud['hrsStudy'] = df_stud['hrsStudy'].astype(str).astype(int)
```

```python
sns.boxplot(df_stud['FinalGrade'])
```

```python
z_scores = stats.zscore(df_stud['FinalGrade'])
abs_z_scores = np.abs(z_scores)
outliers = abs_z_scores > 3
print(df_stud[outliers])
```

```python
df_stud.drop(df_stud[outliers].index, inplace=True)
```

```python
sns.boxplot(df_stud['FinalGrade'])
```

```python
df_stud.info()
```

```python
df_stud = df_stud.reset_index(drop=True)
```

```python
df_stud.head(10)
```

```python
sns.boxplot(df_stud['Age'])
```

```python
z_scores_age = stats.zscore(df_stud['Age'])
abs_z_scores_age = np.abs(z_scores_age)
outliers_age = abs_z_scores_age > 3
print(df_stud[outliers_age])
```

```python
df_stud.drop(df_stud[outliers_age].index, inplace=True)
```

```python
sns.boxplot(df_stud['Age'])
```

```python
df_stud.isna().sum()
```

```python
df_stud = df_stud.reset_index(drop=True)
df_stud
```

```python

# Define bins and labels:
# right=True => intervals are (a, b]; include_lowest=True => lowest bound included in first bin
bins   = [0, 50, 60, 70, 80, 90, 100]
labels = ['F', 'E', 'D', 'C', 'B', 'A']

df_stud['Grade'] = pd.cut(
    df_stud['FinalGrade'],
    bins=bins,
    labels=labels,
    right=True,            # (a, b]
    include_lowest=True    # include the very first bound
).astype('string')
```

```python
grade_order = ['A', 'B', 'C', 'D', 'E', 'F']
df_stud['Grade'] = pd.Categorical(df_stud['Grade'], categories=grade_order, ordered=True)
sns.histplot(df_stud['Grade'])
```

```python
df_stud.head()
```

```python
df_stud.groupby('Grade').count()
```

```python
age_labels = ['<20', '20-30', '>30']
df_stud['AgeGroups'] = pd.cut(df_stud['Age'], bins=3, labels=age_labels)
```

```python
sns.kdeplot(data=df_stud, x='FinalGrade', hue='AgeGroups', multiple='stack')
```

```python
sns.scatterplot(data=df_stud, x='FinalGrade', y='Age', hue='AgeGroups')
```
